```python
import geopandas as gpd
import matplotlib.pyplot as plt

# Use the downloaded address GeoPackage.
addresses = gpd.read_file("https://crowdshipping-simulation.fra1.cdn.digitaloceanspaces.com/riga_addresses.gpkg")
addresses = addresses.to_crs(4326) # Transform from the Latvian standard.
addresses["longitude"] = addresses.geometry.x
addresses["latitude"] = addresses.geometry.y
```

```python
addresses.plot(alpha=0.012, color='r')
plt.title('Riga addresses')
plt.show()
```
